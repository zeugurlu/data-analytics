# Import Libraries

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

# Spacy default stop words

```python
print(nlp.Defaults.stop_words)
```

```python
len(nlp.Defaults.stop_words)
```

# Check if the word is a stop word

```python
nlp.vocab['should'].is_stop
```

```python
nlp.vocab['champion'].is_stop
```

# Adding a stop word

```python
nlp.Defaults.stop_words.add('asap')
```

```python
nlp.vocab['asap'].is_stop
```

# Removing stop word

```python
nlp.Defaults.stop_words.remove('another')
```

```python
# Remove the stop_word tag from the lexeme
nlp.vocab['another'].is_stop = False
```

```python
nlp.vocab['another'].is_stop
```

# Filtering stop words

```python
context = "Weather is good. We are learning NLP. This is a good course"
```

```python
doc = nlp(context)
```

```python
list(doc)
```

```python
for sent in doc.sents:
  print(sent)
```

```python
for sent in doc.sents:
  for token in sent:
    if token.text not in nlp.Defaults.stop_words:
      print(token)
```

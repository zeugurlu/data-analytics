# Import Libraries

```python
import numpy as np
```

# Create the dataset

```python
document = "I am happy. I am learning Natural Language Processing. I am at College"
```

# Converting Text into indices

```python
idx = 1 # This will act as our current index as we loop through the text.
word2idx = {'<unk>': 0} # Initialize word to index dictionary with one element which will map the unknown token to zero.
                        # This will not be used for train set but for the test set if the test set has words that are not
                        # present in the test set.
```

# Start filling Word2idx dictionary

```python
tokens = document.lower().split()
for token in tokens:
  if token not in word2idx:
    word2idx[token] = idx
    idx += 1
```

# Converting Text into numbers

```python
doc_int = []
tokens = document.lower().split()
seq = [word2idx[token] for token in tokens] # list comprehension to map each token to its corresponding text
doc_int.append(seq)
```

```python
print(doc_int)
```

# Craete a text data and convert it into numbers

```python
test = " I am hungry"
```

```python
test_int = []
tokens = test.lower().split()
test_seq = [word2idx.get(token, 0) for token in tokens] # if word not found in dictionary then it will assign "0" index
test_int.append(test_seq)
```

```python
print(test_int)
```

# Word to index mapping with frequency counter

```python
word2idx_freq = {}
tokens = document.lower().split()
for i, token in enumerate(tokens):
  word2idx_freq[token] = word2idx_freq.get(token, {'freq': 0, 'index(s)': []})
  word2idx_freq[token]['freq'] += 1
  word2idx_freq[token]['index(s)'].append(i)
```

```python
print(word2idx_freq)
```

```python
for key, value in word2idx_freq.items():
    print(key, '\t', value)
```

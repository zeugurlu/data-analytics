# Import Librarires

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer ,CountVectorizer
import pandas as pd
```

# Create a text

```python
Document1= 'It is rain today.'
Document2= 'No rain on Sunday.'
Document3= 'There is rain on Friday.'
text = [Document1 , Document2 , Document3]
print(text)
```

```python
#text = ['Hello my name is Zeeshan.','This is Natural language processing class.', 'This class is on Tuesday.']
```

# Instantiate Count Vectorizer and TD-IDF Vectorizer (Without Normalization)

```python
countvectorizer = CountVectorizer()
tfidfvectorizer = TfidfVectorizer(norm = None) # Without Normalization. Also stop_words = None.
```

# Fitting Count Vectorizer and TD-IDF Vectorizer on text

```python
count_matrix = countvectorizer.fit_transform(text)
tfidf_matrix = tfidfvectorizer.fit_transform(text)
```

```python
print(count_matrix)
```

```python
print(countvectorizer.vocabulary_)
```

```python
print(tfidf_matrix)
```

```python
print(tfidfvectorizer.vocabulary_)
```

```python
count_tokens = countvectorizer.get_feature_names_out()
tfidf_tokens = tfidfvectorizer.get_feature_names_out()
```

```python
df_countvect = pd.DataFrame(data = count_matrix.toarray(),index = ['Doc_0','Doc_1', 'Doc_2'],columns = count_tokens)
```

```python
df_tfidfvect = pd.DataFrame(data = tfidf_matrix.toarray(),index = ['Doc_0','Doc_1','Doc_2'],columns = tfidf_tokens)
```

```python
print("Count Vectorizer\n")
print(df_countvect)
```

```python
print("\nTD-IDF Vectorizer\n")
print(df_tfidfvect)
```

```python
# Here , we can see clearly that Count Vectorizer give number of frequency with respect to index of vocabulary
# where as tf-idf consider overall documents of weight of words.
```

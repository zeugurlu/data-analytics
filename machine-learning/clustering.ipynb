```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('../data/E-COM_SYNTH.csv')
df
```

```python
inertia_values = []
training_rage = range(1,14)

for i in training_rage:
    model = KMeans(n_clusters=i, random_state=77)
    model.fit(df)
    inertia_values.append(model.inertia_)

plt.plot(training_rage, inertia_values, marker='o')
```

```python
final_model = KMeans(n_clusters=4, random_state=77)
final_model.fit(df)
```

```python
df["clusters"] = final_model.labels_
df
```
